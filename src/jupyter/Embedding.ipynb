# Embedding（词嵌入层）

```python
import numpy as np
```

```python
class Embedding:
    def __init__(self, vocab_size, embed_dim, max_seq_len):
        print("Embedding layer create successfully!")
        self.token_embed = np.random.randn(vocab_size, embed_dim) * 0.02
        print("词汇表创建成功，我的形状：", self.token_embed.shape)
        self.pos_embed = np.random.randn(max_seq_len, embed_dim) * 0.02
        print("位置编码表创建成果，我的形状：", self.pos_embed.shape)

    def forward(self, token_ids):
        seq_len = token_ids.shape[-1]
        print("句子长度：", seq_len)
        tok_emb = self.token_embed[token_ids]
        print("生成的tok_emb：", tok_emb.shape)
        pos_emb = self.pos_embed[:seq_len]
        print("生成的pos_emb：", pos_emb.shape)
        return tok_emb + pos_emb
```

```python
if __name__ == "__main__":
# 超参数：自建词表一共10个token，id范围0~9
    vocab_size = 10
    embed_dim = 16
    max_seq_len = 32
    embedding_layer = Embedding(vocab_size, embed_dim, max_seq_len)

    # ====== 自创词表（10个单词，手动定义word -> id映射）======
    vocab_list = ["good", "love", "work", "gpt", "models", "I", "great", "ai", "code", "building"]
    word2id = {word: idx for idx, word in enumerate(vocab_list)}
    print("\n词表（vocabulary）:", word2id)

    # 原始英文句子保持不变
    sentence = "I love building gpt models"
    words = sentence.split()
    print("\n输入句子：", sentence)
    print("分词结果（tokenization）：", words)

    # 从自建词表中查找对应的token id
    token_ids = np.array([[word2id[w] for w in words]])
    print("\nToken IDs：", token_ids)
    print(f"token_ids shape: {token_ids.shape}")
    # Embedding前向计算
    emb_out = embedding_layer.forward(token_ids)
    print(f"\nEmbedding输出shape (batch, seq_len, embed_dim): {emb_out.shape}")
    print("Embedding层输出：")
    print(emb_out)
```
