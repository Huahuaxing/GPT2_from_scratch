# FeedForward（FFN，前馈网络层）
放在 Transformer/GPT 解码器的**多头注意力之后**，**对每一个 token 向量单独做非线性变换。**  
![FeedForward](../../image/FeedForward.png)

```python
import numpy as np
```

```python
class FeedForward:
    def __init__(self, embed_dim, ff_dim):
        self.W1 = np.random.randn(embed_dim, ff_dim) * 0.02
        self.b1 = np.zeros(ff_dim)
        self.W2 = np.random.randn(ff_dim, embed_dim) * 0.02
        self.b2 = np.zeros(embed_dim)

    def forward(self, x):
        h = x @ self.W1 + self.b1
        h = np.maximum(0, h)
        return h @ self.W2 + self.b2
```

```python
if __name__ == "__main__":
    # 超参，和你前面GPT项目保持一致 embed_dim=16
    batch_size = 1
    seq_len = 5
    embed_dim = 16
    ff_dim = 64 # 经典4倍放大

    # 实例化前馈网络
    ff = FeedForward(embed_dim=embed_dim, ff_dim=ff_dim)

    # 构造模拟输入：模拟MHA输出，shape [batch, seq_len, embed_dim]
    x = np.random.randn(batch_size, seq_len, embed_dim) * 0.02
    print("输入x shape:", x.shape)

    # 前向计算
    out = ff.forward(x)
    print("FFN输出 out shape:", out.shape)

    # 断言校验：输出形状必须和输入完全一致
    assert out.shape == x.shape, f"形状错误！输入{x.shape},输出{out.shape}"
    print("\n✅ 形状校验通过！")

    # 打印部分数据观察
    print("\n第一个token的输入向量：\n", x[0,0])
    print("\n第一个token经过FFN后的输出向量：\n", out[0,0])
```
